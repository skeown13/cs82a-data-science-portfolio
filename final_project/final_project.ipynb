The question that I am investigating is when it comes to pokemon, does an increase in weight mean an increase in attack power or defensive abilities?

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("pokemon_complete_stats.csv")

display(df.head())
display(df.shape)
display(df.describe())
```

```python
print(df.isna().sum())
```

Cleaning of this dataset is unnecessary as the missing values are not located in the columns that I need for my analysis. 

```python
df["weight_lbs"] = df["weight_hg"] * 0.220462

plt.figure(figsize=(10, 6))

sns.scatterplot(data=df, x="weight_lbs", y="attack", alpha=0.5, edgecolor=None)

sns.regplot(data=df, x="weight_lbs", y="attack", scatter=False, color="red")

plt.title("Entry Weight vs. Attack Power")
plt.xlabel("Weight (lbs)")
plt.ylabel("Attack Power")
plt.show()
```

```python
df["weight_lbs"] = df["weight_hg"] * 0.220462

filtered_df = df[df["weight_lbs"] < 200]

plt.figure(figsize=(10, 6))

sns.scatterplot(data=filtered_df, x="weight_lbs", y="attack", alpha=0.5, edgecolor=None)

sns.regplot(data=filtered_df, x="weight_lbs", y="attack", scatter=False, color="red")

plt.title("Entry Weight vs. Attack Power")
plt.xlabel("Weight (lbs)")
plt.ylabel("Attack Power")
plt.show()
```

Looking at all rows of this dataset it appears that overall there is a positive correlation between a Pokemon's weight and it's attack power, however it is a weak one. When zooming in on the pokemon that weigh less than 200 lbs, as this appears to be where the majority of them weigh, it is confirmed that you can have pokemon with the same relative weight having either a high or low attack power. 

```python
df["weight_lbs"] = df["weight_hg"] * 0.220462

plt.figure(figsize=(10, 6))

sns.scatterplot(data=df, x="weight_lbs", y="defense", alpha=0.5, edgecolor=None)

sns.regplot(data=df, x="weight_lbs", y="defense", scatter=False, color="red")

plt.title("Entry Weight vs. Defense Ability")
plt.xlabel("Weight (lbs)")
plt.ylabel("Defense Ability")
plt.show()
```

```python
df["weight_lbs"] = df["weight_hg"] * 0.220462

filtered_df = df[df["weight_lbs"] < 200]

plt.figure(figsize=(10, 6))

sns.scatterplot(data=filtered_df, x="weight_lbs", y="defense", alpha=0.5, edgecolor=None)

sns.regplot(data=filtered_df, x="weight_lbs", y="defense", scatter=False, color="red")

plt.title("Entry Weight vs. Defensive Ability")
plt.xlabel("Weight (lbs)")
plt.ylabel("Defensive Ability")
plt.show()
```

Interestingly with the correlation between weight of pokemon and their defensive ability, The heaviest of pokemon seem to have a lower defensive stat in relation to the trend line. Zooming in on the pokemon that weigh under 200 lbs, it appears that the weights vs defensive ability seem to be pretty evenly scattered along the trend line. 

In conclusion, for both the comparisons of the Weight of a pokemon to its Attack Power and its Defensive Abilities, there is a slight positive correlation. The trend line does move in the upwards direction. Taking a look at all the scatter plots, it is a weak correlation, but it is there. 

If a person were to want to know more specifics in order to choose the strongest or most resilient pokemon to add to their team, more research would need to be done. Basing that decision solely on weight would not guarantee expected results.
