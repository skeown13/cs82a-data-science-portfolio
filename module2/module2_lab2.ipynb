```python
def multiply_for(num, times):
    result = 0
    for i in range(times):
        result += num
    return result

print(multiply_for(4, 2))
```

```python
def multiply_while(num, times):
    i = 0
    result = 0
    while i < times:
        result += num
        i+=1
    return result

print(multiply_while(5, 3))
```
