```python
def multiply_for(num, times):
    result = 0
    for i in range(times):
        result += num
    return result

print(multiply_for(4, 2))
```

```python
def multiply_while(num, times):
    i = 0
    result = 0
    while i < times:
        result += num
        i+=1
    return result

print(multiply_while(5, 3))
```

A variable is created as a counter variable. At the end of each while loop that counter variable is incremented by 1. Once the i variable is an equal number to the variable times, the loop terminates and the result is returned.
