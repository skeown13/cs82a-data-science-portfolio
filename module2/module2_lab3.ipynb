My Cutoffs for temperatures are as follows:
"COLD" if temperature is below 15
"WARM" if temperature is between 15 and 25 (inclusive)

```python
temp = [10, 16, 23, 28, 14, 22, 30]

for i in temp:
    if i < 15:
        print(f"Today the temperature is {i}, it is COLD.")
    elif i >= 15 and i <= 25:
        print(f"Today the temperature is {i}, it is WARM.")
    else:
        print(f"Today the temperature is {i}, it is HOT.")
```
