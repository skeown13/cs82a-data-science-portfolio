```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

df = pd.read_csv("tips.csv")

X = df[["total_bill"]]
y = df["tip"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42)
len(X_train)
```

We grade on data that the model has never seen so that we can know if the module is actually "learning". If we grade on what it has seen, it already knows this information and it is not a fair assessment. By grading on new information we get an unbiased assessment. 

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train, y_train)
print(model.score(X_test, y_test))
print(model.coef_, model.intercept_)
```

Each extra dollar of bill adds about 0.11 of tip.

```python
y_pred = model.predict(X)

# Plot the scatter points and fitted line
plt.figure(figsize=(8, 5))
plt.scatter(X['total_bill'], y, label='Actual tips')
plt.plot(X['total_bill'], y_pred, color='red', label='Fitted line')

# Fully label the plot
plt.xlabel('Total Bill ($)')
plt.ylabel('Tips ($)')
plt.title('Total Bill vs. Tips with Fitted Regression Line')
plt.legend()
plt.grid(True)
plt.show()
```

```python
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)
print(iris.keys())

X = iris.data
y = iris.target
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42)

len(X_train)
```
