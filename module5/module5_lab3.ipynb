```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv("custdata.tsv", sep="\t")
```

```python
variables = df[["age", "income", "num.vehicles"]]

correlations = variables.corr(method="pearson")

sns.heatmap(
    correlations,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Matrix")
plt.show()
```

```python
df_clean = df.dropna(subset=["num.vehicles"])
df_clean = df_clean[(df_clean["age"] >= 15) & (df_clean["age"] <= 100)]
df_clean = df_clean[df_clean["income"] >= 0]

display(df_clean.shape)
```

```python
variables = df_clean[["age", "income", "num.vehicles"]]

correlations = variables.corr(method="pearson")

sns.heatmap(
    correlations,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Matrix")
plt.show()
```

In cleaning my data, I focused on 3 items. First, I dropped any row that had a value in num.vehicles of na; due to not having complete data. Second, I removed any row where the age was less than 15 or older than 100; in our society for the most part you don't have people that are these ages working/gaining an income. Finally, I removed any row where the income was a negative number; you can't have negative income.

From my data, what I can tell is that there is not much of a correlation between age, income, and number of vehicles. The strongest correlation that I see is between income and number of vehicles. It appears that as one increases so does the other; though not by much.
