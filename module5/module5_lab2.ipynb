```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("custdata.tsv", sep="\t")
```

```python
display(df.shape)
df.isna().sum()
```

```python
df_clean = df.dropna(subset=["housing.type"])
display(df_clean.shape)
df_clean.isna().sum()
```

```python
counts = df_clean["housing.type"].value_counts()

print(counts)

counts.plot(kind="bar", color="lime", edgecolor="black")

plt.title("Customer's Type of Housing")
plt.xlabel("Housing Type")
plt.ylabel("Number of Customers")
plt.xticks(rotation=45, ha="right")
plt.tight_layout() 
plt.show()
```

I excluded 56 rows from the dataset due to their housing type being na.
