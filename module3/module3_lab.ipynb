```python
import pandas as pd
df = pd.read_csv("messy_sales.csv")
display(df.shape)
display(df.head(10))
display(df.dtypes)
```

## 3 Problems I see with the above data:

- data type for "date" is reading as an object
- data type for "zip" is reading as an integer
- there are values of NaN for some prices

```python
df.isna().sum()
```

```python
fill_value = df["price"].median()
df["price"] = df["price"].fillna(fill_value)
print(fill_value)
```

```python
df.isna().sum()
```

```python
print(df.duplicated().sum())
df = df.drop_duplicates()
df.shape
```

```python
df["zip"] = df["zip"].astype(str).str.zfill(5)
```

```python
df["date"] = pd.to_datetime(df["date"], format="mixed")
display(df.dtypes)
```

```python
display(df.shape)
display(df.head(10))
display(df.dtypes)
```

```python
import numpy as np
df[df["qty"] < 0]
df["neg_qty"] = np.where(df["qty"] < 0, 1, 0)
df[df["qty"] < 0]
```

```python
df.to_csv('sales_clean.csv', index = False)
```
