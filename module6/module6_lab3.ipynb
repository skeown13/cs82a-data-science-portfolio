```python
import pandas as pd
import matplotlib.pyplot as plt
import sqlite3

con = sqlite3.connect("customers.db")
con.execute("DROP TABLE IF EXISTS Customers")
con.execute('''
CREATE TABLE 'Customers' (
'ID' INTEGER PRIMARY KEY,
'Name' TEXT NOT NULL default '',
'City' TEXT NOT NULL default '',
'Age' INTEGER NOT NULL default 0
);
''')

con.execute('''
INSERT INTO Customers (Name, City, Age) VALUES
('Billy', 'Los Angeles', 47),
('Susan', 'Houston', 33),
('Jerry', 'San Francisco', 21),
('Darlene', 'Los Angeles', 22),
('Jessica', 'Loveland', 37),
('Christopher', 'Cincinnati', 2),
('Lacey', 'Winchester', 13),
('Caspian', 'Westchester', 42);
''')

con.commit()
```

```python
pd.read_sql('''
SELECT Name 
FROM Customers
WHERE City = 'Los Angeles'
''', con)
```

```python
desc_age = pd.read_sql('''
SELECT *
FROM Customers
ORDER BY Age DESC
''', con)

print(desc_age)
```

```python
dbl_letter_name = pd.read_sql('''
SELECT Name, Age
FROM Customers
WHERE Name LIKE '%aa%'
or Name LIKE '%bb%'
or Name LIKE '%cc%'
or Name LIKE '%dd%'
or Name LIKE '%ee%'
or Name LIKE '%ff%'
or Name LIKE '%gg%'
or Name LIKE '%hh%'
or Name LIKE '%ii%'
or Name LIKE '%jj%'
or Name LIKE '%kk%'
or Name LIKE '%ll%'
or Name LIKE '%mm%'
or Name LIKE '%nn%'
or Name LIKE '%oo%'
or Name LIKE '%pp%'
or Name LIKE '%qq%'
or Name LIKE '%rr%'
or Name LIKE '%ss%'
or Name LIKE '%tt%'
or Name LIKE '%uu%'
or Name LIKE '%vv%'
or Name LIKE '%ww%'
or Name LIKE '%xx%'
or Name LIKE '%yy%'
or Name LIKE '%zz%'
''', con)

print(dbl_letter_name)
```

```python
ax = desc_age.plot(x="Name", y="Age", kind="bar", legend=False)
# sns.barplot(data="df_city", x="city", y="total_spend")
plt.title("Age of Citizens")
plt.xlabel("Citizen Name")
plt.ylabel("Age")

ax.bar_label(ax.containers[0], fmt='%g', padding=3)
plt.ylim(0, desc_age["Age"].max() * 1.1)

plt.show()
```
