```python
import pandas as pd
df = pd.read_csv("truck.csv")
```

```python
df.corr(numeric_only=True)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create a scatter plot with custom axis labels
ax = df.plot.scatter(x="X", y="Y")
ax.set_xlabel('Listed Price (in thousands)')
ax.set_ylabel('Best Price (in thousands)')
plt.title("Truck's Listed Price vs Best Price")
plt.show()
```

The trends of what a truck's list price is vs what its best price is, is quite linear in its progression. The smaller listed priced trucks have a difference from the best price of about 1.8k and steadily increases to a differnce of 2.6k for the higher listed prices. With this knowledge I can say, with absolute certainty, that if a truck has a list price of 22.8k that the best price for it will be 20.2k.
