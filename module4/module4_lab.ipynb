```python
import pandas as pd
df = pd.read_csv("./../module3/sales_clean.csv")
del df["neg_qty"]
display(df.shape)
```

```python
print(df["price"].mean(), "mean price")
print(df["price"].median(), "median price")
print(df["price"].std(), "std price")
print(df["qty"].mean(), "mean qty")
print(df["qty"].median(), "median qty")
print(df["qty"].std(), "std qty")
```

I would report the median for price. Since the mean price is a decent amount higher than the median price, the data is skewed. So there is at least one entry that is much higher than the rest pulling the mean higher. Hence utilizing the median is better in this scenario as it sits at the halfway point in the data. 

```python
df["price"].plot(kind="hist", bins=20)

# The histogram shape is right-skewed
```

```python
df["revenue"] = df["price"] * df["qty"]
df.head()
```

```python
df.corr(numeric_only=True)
```

```python
df.plot.scatter(x="qty", y="revenue")
```

The relationship between qty and revenue appears to be that the more qty the higher the revenue. The strength of this seems accurate due to the fact that if more items are being bought, more money will be gained. A cautionary outlier would be that if an item is more expensive than a quantity of 1 may yield the same in revenue of a quantity of say 3 cheaper items.

Compute the mean and median of [2, 4, 4, 6, 9] by hand.
Mean: The 5 numbers added together is 25. Then divide 25 by 5 and get a mean of 5.
Median: The center number when the numbers are ordered smallest to largest; makes the median equal to 4.

```python
import numpy as np
my_data = [2, 4, 4, 6, 9]
print(np.mean(my_data))
print(np.median(my_data))
```
