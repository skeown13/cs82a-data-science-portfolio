```python
# Weights of Books

print(df["weight"].mean(), "This is the average weight of all books recorded regardless of their cover type.")
mean_weight_hb = df[df["cover"] == "hb"]["weight"].mean()
print(mean_weight_hb, "This is the average weight of all the books with a hardback cover type.")
mean_weight_pb = df[df["cover"] == "pb"]["weight"].mean()
print(mean_weight_pb, "This is the average weight of all the books with a paperback cover type.")
```

The difference between the mean weights is that the hb books weigh more on average than the pb books.
